## 2.1 Dasar Teori
 Perulangan for merupakan salah satu struktur Python yang digunakan untuk menjalankan perintah secara berulang. Dengan for, program dapat memproses setiap elemen dalam data atau melakukan perulangan dengan jumlah tertentu tanpa menulis perintah yang sama berulang kali.

1. Pernyataan for digunakan untuk mengulang perintah pada setiap elemen data, seperti list, string, atau tuple. Perulangan berhenti setelah semua elemen selesai diproses.
2. for dengan range() digunakan ketika jumlah perulangan sudah ditentukan. range() dapat menentukan nilai awal, nilai akhir, dan jarak antar angka.
3. Nested for merupakan perulangan for yang berada di dalam perulangan lainnya. Biasanya digunakan untuk membuat pola, tabel perkalian, atau mengolah data dalam bentuk baris dan kolom.

## 2.2 Kode Program

### 2.2.1 Program perulangan_for1__2611531021


```python
ulang_3021 = int(input("Masukkan jumlah pengulangan: "))

for i_3021 in range(ulang_3021):
    print(f"perulangan ke-{i_3021}")
```

**Penjelasan :**
1. ulang_3021 = int(input("masukkan jumlah perulangan: "))
Meminta pengguna memasukkan jumlah perulangan. Nilai input yang awalnya berupa string diubah menjadi bilangan bulat menggunakan int() dan disimpan dalam variabel ulang_3021.
2. for i_3021 in range(ulang_3021):
Menjalankan perulangan sesuai nilai ulang_3021. Fungsi range() menghasilkan angka mulai dari 0 hingga sebelum nilai tersebut, yang kemudian disimpan dalam i_3021.
2. print(f"perulangan ke-({i_3021})")
Menampilkan nomor perulangan pada setiap proses. Huruf f memungkinkan nilai variabel i_3021 dimasukkan langsung ke dalam teks menggunakan {}.

### 2.2.3 Program Perulangan_for3_2611533021

```python
ulang_3021 = int(input("Masukkan jumlah_3021 pengulangan: "))

jumlah_3021 = 0
for i_3021 in range(1, ulang_3021 + 1):
    print(i_3021, end=" ")
    jumlah_3021 = jumlah_3021 + i_3021

    if i_3021 < ulang_3021:
        print("+", end=" ")
    else:
        print("= ", jumlah_3021,end=" ")
print()
print("jumlah =", jumlah_3021)
```

**Penjelasan :**
1. Input jumlah perulangan
Program meminta pengguna memasukkan angka, kemudian mengubahnya menjadi integer dan menyimpannya dalam ulang_3021.
2. Inisialisasi penjumlahan
Variabel jumlah_3021 digunakan untuk menyimpan total penjumlahan dan dimulai dari nilai 0.
3. Perulangan for
Perulangan berjalan dari 1 hingga angka yang dimasukkan. +1 digunakan karena batas akhir pada range() tidak ikut dihitung.
4. Proses penjumlahan
print(i_3021, end=" ") menampilkan angka dalam satu baris, sedangkan jumlah_3021 = jumlah_3021 + i_3021 menambahkan setiap angka ke total. Pernyataan if digunakan untuk mengatur tampilan tanda + dan hasil akhir.
5. Output akhir
print() digunakan untuk berpindah ke baris baru, kemudian program menampilkan hasil total penjumlahan.

### 2.2.4 Program nested_for1_2611533021

```python
batas_3021 = int(input("Masukkan nilai batas: "))
for line_3021 in range(1, batas_3021 + 1):
    for j_3021 in range(1, (-1 * line_3021 + batas_3021 + 1)):
        print(".", end="")
    print(line_3021)
```

**Penjelasan :**
1. batas_3021 = int(input("Masukkan nilai batas: "))
Meminta pengguna memasukkan nilai batas, kemudian mengubahnya menjadi bilangan bulat. Nilai ini menentukan jumlah baris yang dibuat.
2. for line_3021 in range(1, batas_3021+1):
Merupakan perulangan luar yang digunakan untuk menentukan setiap baris, mulai dari 1 hingga nilai batas.
3. for j_3021 in range(1, (-1 * line_3021 + batas_3021)+1):
Merupakan perulangan dalam yang menentukan jumlah titik pada setiap baris. Semakin besar nomor baris, semakin sedikit titik yang ditampilkan.
4. print(".", end=" ")
Menampilkan titik dan spasi tanpa berpindah ke baris baru, sehingga titik tetap berada dalam satu baris.
5. print(line_3021)
Menampilkan nomor baris setelah semua titik selesai dicetak, kemudian berpindah ke baris berikutnya.

### 2.2.7 Program nested_for4_2611533021

```python
tinggi_3021 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3021 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3021 = tinggi_3021
    c_3021 = a_3021
    lebar_3021 = (2 * tinggi_3021) - 2

    for i_3021 in range(1, tinggi_3021 + 1):
        b_3021 = c_3021 + 1

        for j_3021 in range(1, lebar_3021 + 1):

            # Baris atas dan bawah
            if i_3021 == 1 or i_3021 == tinggi_3021:
                if j_3021 == 1 or j_3021 == lebar_3021:
                    print("#", end="")
                else:
                    print("=", end="")
            # Baris isi
            else:
                if j_3021 == 1 or j_3021 == lebar_3021:
                    print("|", end="")
                else:
                    if j_3021 == c_3021:
                        print("<", end="")
                    elif j_3021 == b_3021:
                        print(">", end="")
                    elif j_3021 == (lebar_3021 - c_3021):
                        print("<", end="")
                    elif j_3021 == (lebar_3021 - c_3021 + 1):
                        print(">", end="")
                    elif j_3021 > b_3021 and j_3021 < (lebar_3021 - c_3021):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        #Logika asli java
        a_3021 -= 2

        if a_3021 <= 0:
            c_3021 = (-a_3021) + 2
        else:
            c_3021 = a_3021
```

**Penjelasan :**
1. Input dan validasi
Pengguna memasukkan tinggi pola. Jika nilai yang dimasukkan ganjil, program menampilkan pesan error. Jika genap, program melanjutkan proses.
2. Inisialisasi variabel
Variabel a_3021 dan c_3021 diatur sesuai tinggi pola. Variabel c menentukan posisi tanda <, sedangkan lebar_3021 = 2 * tinggi - 2 menentukan lebar pola.
3. Perulangan luar (baris)
Perulangan berjalan dari 1 hingga nilai tinggi. Pada setiap baris, posisi tanda > ditentukan berdasarkan nilai c.
4. Perulangan dalam (kolom)
Setiap kolom diperiksa untuk menentukan karakter yang akan dicetak, seperti #, =, |, <, >, ., atau spasi sesuai posisi masing-masing.
5. Pembaruan pola
Setelah satu baris selesai, program berpindah ke baris berikutnya dan memperbarui nilai c. Posisi tanda < dan > bergerak ke arah tepi lalu kembali ke tengah, sehingga menghasilkan pola yang simetris.
6. Hasil akhir
Jika tinggi yang dimasukkan adalah 10, program menghasilkan pola berbentuk simetris menyerupai heksagon atau belah ketupat yang melebar di bagian tengah dan menyempit kembali di bagian bawah.

Perulangan berjalan dari 1 hingga nilai batas yang telah dimasukkan.
4. Seleksi bilangan genap
Program hanya memproses angka yang habis dibagi 2, sedangkan angka ganjil dilewati.
5. Proses penjumlahan
Setiap bilangan genap ditampilkan dalam satu baris dan ditambahkan ke jumlah_3021. Tanda + atau = digunakan untuk menampilkan bentuk perhitungannya.
6. Output akhir
Setelah perulangan selesai, program berpindah ke baris baru dan menampilkan hasil total penjumlahan. Misalnya, jika batasnya 10, hasilnya adalah 2 + 4 + 6 + 8 + 10 = 30.

## Tugas Praktikum

**Studi Kasus:** Pola Jam Pasir Kristal Palindromik Berbingkai (Framed Palindromic Crystal Hourglass)

**Kode Jawaban :**

```python
print("==== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ====")
x_3021 = int(input("Masukkan ukuran skala jam pasir (N): "))

# ---------- Bingkai atas ----------
print("#", end="")
for garis_3021 in range(4 * x_3021 + 5):
    print("=", end="")
print("#")

# ---------- Fase 1: jam pasir atas (baris N turun s.d. 1) ----------
for baris_3021 in range(x_3021, 0, -1):
    print("|", end="")
    print(" ", end="")
    # spasi penyeimbang kiri
    for spasi_kiri_3021 in range(2 * (x_3021 - baris_3021)):
        print(" ", end="")
    # deret angka mundur (baris -> 1)
    for angka_mundur_3021 in range(baris_3021, 0, -1):
        print(angka_mundur_3021, end=" ")
    # poros kristal
    print("<*>", end="")
    # deret angka maju (1 -> baris)
    for angka_maju_3021 in range(1, baris_3021 + 1):
        print(" ", end="")
        print(angka_maju_3021, end="")
    # spasi penyeimbang kanan
    for spasi_kanax_3021 in range(2 * (x_3021 - baris_3021)):
        print(" ", end="")
    print(" ", end="")
    print("|", end="")
    print()

# ---------- Fase 2: poros titik pusat ----------
print("|", end="")
for spasi_kiri_3021 in range(2 * x_3021 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_kanax_3021 in range(2 * x_3021 + 1):
    print(" ", end="")
print("|", end="")
print()

# ---------- Fase 3: jam pasir bawah (baris 1 naik s.d. N) ----------
for baris_3021 in range(1, x_3021 + 1):
    print("|", end="")
    print(" ", end="")
    # spasi penyeimbang kiri
    for spasi_kiri_3021 in range(2 * (x_3021 - baris_3021)):
        print(" ", end="")
    # deret angka mundur (baris -> 1)
    for angka_mundur_3021 in range(baris_3021, 0, -1):
        print(angka_mundur_3021, end=" ")
    # poros kristal
    print("<*>", end="")
    # deret angka maju (1 -> baris)
    for angka_maju_3021 in range(1, baris_3021 + 1):
        print(" ", end="")
        print(angka_maju_3021, end="")
    # spasi penyeimbang kanan
    for spasi_kanax_3021 in range(2 * (x_3021 - baris_3021)):
        print(" ", end="")
    print(" ", end="")
    print("|", end="")
    print()

# ---------- Bingkai bawah ----------
print("#", end="")
for garis_3021 in range(4 * x_3021 + 5):
    print("=", end="")
print("#")
```

**Penjelasan:**

1. Penamaan variabel

Semua variabel menggunakan akhiran _3021 sesuai ketentuan program.

2. Membuat bingkai

Program mencetak karakter # di awal dan akhir, serta karakter = di antaranya menggunakan perulangan for tanpa perkalian string.

3. Fase 1 (pola menurun)

Perulangan berjalan dari nilai N hingga 1. Setiap baris menampilkan garis tepi, spasi, angka yang disusun mundur dan maju, serta huruf M di bagian tengah.

4. Fase 2 (baris poros)

Program mencetak pola <*> dengan spasi di sisi kiri dan kanan sebagai bagian tengah pola.

5. Fase 3 (pola menaik)

Pola dicetak kembali seperti fase pertama, tetapi perulangan berjalan dari 1 hingga N, sehingga susunan angka membentuk pola yang simetris.

6. Pindah baris

Perintah print() tanpa argumen digunakan pada akhir setiap baris agar pola tersusun dengan rapi.
