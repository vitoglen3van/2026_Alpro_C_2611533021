## 2.1 Dasar Teori
 If Tunggal: Digunakan saat program hanya perlu mengeksekusi aksi jika suatu kondisi terpenuhi, dan mengabaikannya jika kondisi bernilai False.

If-Else: Struktur percabangan dua jalur. Jika kondisi bernilai True, blok if dieksekusi; jika False, blok else yang dijalankan.

If-Elif-Else: Struktur percabangan multi-jalur mutlak (mutually exclusive). Evaluasi dilakukan secara berurutan dari atas ke bawah. Ketika salah satu kondisi bernilai True, blok kode tersebut dieksekusi dan seluruh sisa blok elif/else di bawahnya diabaikan.

Multi-IF Terpisah: Struktur yang menggunakan beberapa blok if mandiri secara berurutan. Setiap blok if dievaluasi secara independen sehingga memungkinkan pengakumulasian nilai (seperti akumulasi diskon) apabila lebih dari satu kondisi terpenuhi secara bersamaan.

Match-Case: Fitur structural pattern matching yang diperkenalkan pada Python 3.10. Digunakan untuk menggantikan rantai if-elif panjang pada pemilihan nilai diskrit, di mana case _ berfungsi sebagai wildcard (pilihan default).

### 2.2.4 Program multi_if2_2611533021

```python
#input dari user
total_belanja_3021 = float(input("Masukkan total belanja (Rp): "))

#input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3021 = input("Apakah anda member? (y/t): ").strip().lower()
is_member_3021 = input_member_3021 in ["y", "ya"]

#input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3021 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3021 = input_promo_3021 in ["y", "ya"]

total_diskon_persen_3021 = 0

if total_belanja_3021 > 1000000:
    total_diskon_persen_3021 += 10 #Diskon belanja besar

if is_member_3021:
    total_diskon_persen_3021 += 5 #Diskon member

if kode_promo_valid_3021:
    total_diskon_persen_3021 += 15 #Diskon voucher

nominal_diskon_3021 = total_belanja_3021 * (total_diskon_persen_3021 / 100)
total_bayar_3021 = total_belanja_3021 - nominal_diskon_3021

print("\n--- Rincian Pembayaran ---")
print(f"Total diskon    : {total_diskon_persen_3021}% (Rp {nominal_diskon_3021:,.0f})")
print(f"Total bayar     : Rp {total_bayar_3021:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_3021}%")
```

**Penjelasan :**
1. Program meminta pengguna memasukkan total belanja menggunakan input().
2. Nilai total belanja diubah menjadi float dan disimpan dalam variabel total_belanja_3021.
3. Program meminta status member dengan pilihan y atau t, kemudian input diubah menjadi huruf kecil menggunakan .lower() dan dibersihkan dengan .strip().
4. Program memeriksa apakah input member berupa "y" atau "t" menggunakan operator in, lalu menyimpannya dalam is_member_3021.
5. Program meminta status kode promo dengan pilihan y atau t, kemudian hasilnya disimpan dalam kode_promo_valid_3021.
6. Variabel total_diskon_persen_3021 diberi nilai awal 0 untuk menyimpan persentase diskon.
7. Jika total belanja lebih dari Rp1.000.000, program menambahkan diskon 10%.
8. Jika pengguna merupakan member, program menambahkan diskon 5%.
9. Jika kode promo valid, program menambahkan diskon 15%.
10. Setiap kondisi diperiksa secara terpisah sehingga diskon dari beberapa kondisi dapat diakumulasikan.
11. Program menghitung nominal diskon berdasarkan total belanja dan persentase diskon.
12. Program menghitung total pembayaran dengan mengurangi total belanja dengan nominal diskon.

### 2.2.5 Program  if_elif_else1_2611533021

```python
umur_3021 = int(input("Input umur anda: "))
sim_3021 = input("Apakah anda sudah punya sim C (y/t): ")[0]

if umur_3021 >= 17 and sim_3021 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")
elif umur_3021 >= 17 and sim_3021 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
elif umur_3021 < 17 and sim_3021 == 'y':
    print("Anda belum cukup umur untuk punya SIM")
else:
    print("Anda belum cukup umur dan tidak boleh bawa motor")
print("Program Selesai")
```

4. Program memeriksa apakah umur 17 tahun atau lebih dan memiliki SIM (sim_3021 == "y"). Jika benar, pengguna dinyatakan sudah dewasa dan boleh membawa motor.
5. Jika kondisi pertama salah, program memeriksa elif apakah umur 17 tahun atau lebih tetapi tidak memiliki SIM. Jika benar, pengguna dinyatakan sudah dewasa tetapi tidak boleh membawa motor.
6. Jika kondisi kedua salah, program memeriksa elif apakah umur kurang dari 17 tahun tetapi memiliki SIM. Jika benar, program menampilkan bahwa pengguna belum cukup umur untuk memiliki SIM.
7. Jika semua kondisi sebelumnya salah, bagian else dijalankan dan menampilkan bahwa pengguna belum cukup umur dan tidak boleh membawa motor.

### 2.2.5 Program match_case1_2611533021

```python
bulan_3021 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_3021:
    case 1:
        print("Januari")
    case 2:
        print("Februrari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan :**
1. Program meminta pengguna memasukkan angka bulan dari 1 sampai 12 menggunakan input().
2. Nilai diubah menjadi int dan disimpan dalam variabel bulan_3021.
3. Program menggunakan struktur match-case untuk mencocokkan nilai bulan_3021.
4. Jika nilai 1, program menampilkan Januari.
5. Jika nilai 2 sampai 12, program menampilkan nama bulan yang sesuai, dari Februari hingga Desember.
6. Jika angka di luar 1 sampai 12, case _ dijalankan dan menampilkan pesan "Angka tidak valid".
7. Setelah nama bulan atau pesan kesalahan ditampilkan, program selesai.
