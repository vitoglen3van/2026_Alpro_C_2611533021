## 2.1 Dasar Teori
 Penjelasan mengenai konsep dan teori yang menjadi dasar pelaksanaan praktikum ini mencakup beberapa hal berikut.


1. Bahasa Pemrograman Python dan Sintaksis Dasar
 Python merupakan bahasa pemrograman tingkat tinggi yang menggunakan interpreter dan mendukung pemrograman berorientasi objek. Python dikenal karena sintaksisnya yang sederhana dan mudah dibaca, sehingga programmer dapat menuliskan suatu konsep dengan kode yang relatif singkat dibandingkan beberapa bahasa pemrograman lain seperti C++ atau Java (Severance, 2016). Beberapa sintaks dasar Python yang digunakan dalam praktikum antara lain:
 - Output: Fungsi "print()" digunakan untuk menampilkan teks atau data ke layar.
- Penggabungan String: Operator "+" dapat digunakan untuk menggabungkan beberapa teks menjadi satu.
- Escape Sequence: Karakter khusus yang menggunakan tanda backslash ("\"). Contohnya, "\\" digunakan untuk menampilkan karakter backslash secara langsung.
- Komentar: Digunakan untuk memberikan keterangan pada kode dan tidak akan dijalankan oleh Python. Komentar satu baris diawali dengan tanda "#", sedangkan komentar yang lebih dari satu baris dapat ditulis menggunakan tiga tanda kutip.

2. Lingkungan Kerja Pemrograman
Lingkungan kerja pemrograman membantu programmer dalam menulis, menjalankan, menguji, dan memperbaiki kode. Beberapa tools yang dapat digunakan dalam praktikum ini adalah Visual Studio Code, Python Debugger, dan Jupyter (Microsoft, 2023).
- Visual Studio Code (VS Code): Editor kode yang ringan dan menyediakan berbagai fitur serta ekstensi untuk membantu proses pemrograman.
- Python Debugger: Digunakan untuk mencari dan memahami kesalahan dalam program. Programmer dapat menghentikan program pada bagian tertentu menggunakan breakpoint atau menjalankan kode secara bertahap.
- Jupyter: Lingkungan pemrograman interaktif yang memungkinkan kode dijalankan dalam bagian-bagian kecil atau code cell. Hal ini memudahkan proses percobaan dan pengujian kode secara langsung (Kluyver et al., 2016).

3. Pengelolaan Kode dengan Git dan GitHub
Dalam pengembangan program, perubahan kode perlu dikelola dengan baik agar riwayat pekerjaan dapat dilacak dan kesalahan dapat diperbaiki dengan lebih mudah. Untuk kebutuhan tersebut, digunakan Version Control System (VCS), yaitu sistem yang mencatat perubahan pada file dari waktu ke waktu (Chacon & Straub, 2014).
- Git: Sistem pengelolaan versi terdistribusi yang digunakan untuk mencatat dan mengatur perubahan kode secara lokal pada komputer.
- GitHub: Platform yang digunakan untuk menyimpan repositori Git secara daring. GitHub juga membantu dalam mencadangkan kode, melihat riwayat perubahan, serta memungkinkan beberapa orang bekerja sama dalam satu proyek.

### 2.2.2 Program Repository GitHub

Setelah program dibuat dan diuji pada VS Code, proyek dihubungkan ke platform GitHub. Repositori ini berfungsi untuk menyimpan, mengelola, dan mengamankan berkas kode program secara daring (online).

Tahapan yang dilakukan yaitu:

Membuat repositori baru pada halaman GitHub.
Membuka folder proyek menggunakan VS Code.
Membuka terminal terintegrasi pada VS Code.
Menginisialisasi Git pada folder proyek lokal.
Menambahkan berkas ke dalam staging area.
Membuat commit untuk menyimpan riwayat perubahan.
Menghubungkan repositori lokal dengan repositori di GitHub.
Melakukan push untuk mengunggah kode ke GitHub.

Contoh perintah (command) yang digunakan pada terminal:

```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

1. git init: Digunakan untuk membuat dan menginisialisasi repositori Git baru pada folder proyek lokal.
2. git add .: Digunakan untuk menambahkan seluruh berkas dan perubahan pada proyek ke dalam staging area.
3. git commit: Digunakan untuk menyimpan riwayat perubahan dari staging area disertai pesan deskriptif.
4. git branch -M main: Digunakan untuk mengubah atau mengonfirmasi nama cabang (branch) utama menjadi main.
5. git remote add origin: Digunakan untuk menghubungkan repositori lokal dengan repositori yang telah dibuat di GitHub.
6. git push: Digunakan untuk mengunggah berkas dan riwayat commit dari komputer lokal ke repositori GitHub.
