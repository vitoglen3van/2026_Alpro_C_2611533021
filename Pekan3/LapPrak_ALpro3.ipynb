## 1.1 Latar Belakang
Dalam pemrograman, pengolahan data tidak hanya berkaitan dengan cara menyimpan data ke dalam variabel, tetapi juga bagaimana data tersebut diolah dan dimanipulasi untuk menghasilkan keluaran yang sesuai. Salah satu bagian penting yang digunakan dalam proses tersebut adalah operator. Operator merupakan simbol atau instruksi khusus yang digunakan untuk melakukan berbagai operasi terhadap nilai atau variabel (*operand*), seperti perhitungan, perbandingan, maupun pengambilan keputusan berdasarkan kondisi tertentu.

Python menyediakan berbagai jenis operator yang dapat digunakan sesuai dengan kebutuhan program. Operator aritmetika digunakan untuk melakukan perhitungan matematika, sedangkan operator penugasan (*assignment*) digunakan untuk memberikan atau mengubah nilai suatu variabel. Selain itu, terdapat operator perbandingan (*relational*) dan operator logika yang banyak digunakan dalam membuat kondisi dan mengatur alur program. Python juga memiliki operator bitwise untuk mengolah data dalam bentuk biner, operator keanggotaan (*membership*) untuk memeriksa keberadaan suatu nilai dalam kumpulan data, serta operator identitas (*identity*) untuk mengetahui apakah dua variabel merujuk pada objek yang sama.

Memahami berbagai jenis operator tersebut merupakan dasar yang penting dalam mempelajari pemrograman. Penggunaan operator yang kurang tepat dapat menyebabkan *logical error*, yaitu ketika program tetap berjalan tetapi menghasilkan keluaran yang tidak sesuai dengan yang diharapkan. Oleh karena itu, praktikum operator Python dilakukan agar praktikan dapat memahami fungsi dan cara penggunaan setiap jenis operator serta mampu menerapkannya secara langsung dalam program.

## 2.1 Dasar Teori
 Operator adalah simbol atau kata kunci yang digunakan untuk melakukan operasi terhadap satu atau lebih nilai. Nilai yang dioperasikan disebut *operand*, sedangkan gabungan operator dan operand disebut ekspresi. Contohnya, pada `7 + 3`, `+` adalah operator, sedangkan `7` dan `3` adalah *operand*. Berdasarkan jumlah operand, operator dibagi menjadi unary, seperti `-x`, dan binary, seperti `x + y`.

Python memiliki beberapa jenis operator dengan fungsi yang berbeda. Operator aritmetika digunakan untuk perhitungan, seperti `+`, `-`, `*`, `/`, `//`, `%`, dan `**`. Operator perbandingan (`==`, `!=`, `>`, `<`, `>=`, `<=`) digunakan untuk membandingkan nilai dan menghasilkan Boolean. Operator logika (`and`, `or`, `not`) digunakan untuk mengolah kondisi. Operator *bitwise* (`&`, `|`, `^`, `~`, `<<`, `>>`) digunakan untuk operasi pada tingkat bit. Operator identitas (`is`, `is not`) digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama, sedangkan operator keanggotaan (`in`, `not in`) digunakan untuk memeriksa keberadaan nilai dalam struktur data.

Jika beberapa operator digunakan dalam satu ekspresi, Python mengikuti aturan *precedence* atau prioritas operator. Secara umum, urutannya dimulai dari tanda kurung, perpangkatan, unary, perkalian/pembagian, penjumlahan/pengurangan, *bitwise*, perbandingan, kemudian logika. Contohnya, `2 + 3 * 2` menghasilkan `8` karena perkalian memiliki prioritas lebih tinggi daripada penjumlahan.

### 2.2.5 Program Lainnya_2611533021

````python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("======================================")
print("1. OPERATOR KEANGGOTAAN")
print("======================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_3021 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_3021 = [int(angka.strip()) for angka in input_data_3021.split(",")]

nilai_dicari_3021 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_3021 = nilai_dicari_3021 in data_3021
print("\nOperator keanggotaan IN")
print(nilai_dicari_3021,"in",data_3021,"=",hasil_3021)

# Operator not in
hasil_3021 = nilai_dicari_3021 not in data_3021
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_3021,"not in",data_3021,"=",hasil_3021)

print("======================================")
print("2. OPERATOR IDENTIAS")
print("======================================")

# objek1 menggunakan list dari input perngguna
object1_3021 = data_3021

# objek2 merujuk pada objek yang sama dengan objek1
object2_3021 = object1_3021

# objek3 memiliki isi sama, tetapi merupakan objek baru
object3_3021 = data_3021.copy()

print("object1 =",object1_3021)
print("object2 =",object2_3021)
print("object3 =",object3_3021)

# Operator is
hasil_3021 = object1_3021 is object2_3021
print("\nOperator identitas IS")
print("objek1 is objek2 =",hasil_3021)

# Operator is not
hasil_3021 = object1_3021 is not object2_3021
print("\nOperator identitas IS NOT")
print("objek1 is not objek2 =",hasil_3021)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =",object1_3021 is object3_3021)
print("objek1 == objek3 =",object1_3021 == object3_3021)
````
